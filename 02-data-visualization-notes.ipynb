### Data Visualization

Data visualization is the graphical representation of data using visual elements (charts, graphs, maps). It helps communicate relationships among data to viewers, translating complex datasets into a clear story for non-technical audiences. 

Common types of visualizations include:
* Line graphs: Track changes and trends over time.
* Bar charts: Compare data across categories.
* Scatter plots: Show relationships of correlations between two variables.
* Heatmaps: Display data density and regional patterns using color.

### Import libraries

* [Matplotlib](https://matplotlib.org/): A plotting and visualization library for Python. We'll use the `pyplot` module, a collection of functions in the Matplotlib library. By convention, it is imported with the alias `plt`.
* [Seaborn](https://seaborn.pydata.org/): A data visualization library build on top of Matplotplib, providing a high-level interface for creating attractive and complex charts.

### Improving Default Styles using Seaborn

An easy way to make your charts more attractive is to apply default styles provided in the Seaborn library. These can be applied *globally* using the `sns.set_style()` function.

##### **`sns.set_style()`**

The `sns.set_style()` function changes the global visual theme of your plots by altering the background color, grids, and border appearance. It does so by modifying Matplotlib's underlying global configuration settings (`rcParams`).

Syntax: `sns.set_styles(style=None, rc=None)`

Parameters:
* `style`: The name of a preconfigured style, or, a dictionary of parameters
* `rc`: Parameter mappings to override the values in a preconfiured style. 

Seaborn offers five preset themes, which can be passed to the `style` argument:
* `darkgrid` (default): Gray background with white grid lines.
* `whitegrid`: White background with gray grid lines.
* `dark`: Gray background without grid lines.
* `white`: White background with top / right borders, no grid lines. 
* `ticks`: Similar to `white`, but includes tick marks on the axes for precise measurement. 

Let's apply the `darkgrid` style globally, to all charts.

````python
sns.set_style("darkgrid");
````

### Line Chart

A line chart displays information as a series of data points connected by straight lines. It is commonly used to visualize trends, changes, or patterns in a data over a sequence, such as over time. 

The `plt.plot()` function is used to create a line chart. 


##### **`plt.plot()`**

Syntax: `plt.plot(x, y, color=None, linestyle='-', linewidth=1, marker=None, label=None)`

Parameters:
* `x`: values displayed on x-axis
* `y`: values displayed on y-axis
* `color`: color of the line
* `linestyle`: defines the style of the line, such as solid ('-'), dashed ('--'), dotted (':'), dash-dot ('-')
* `linewidth`: sets the thickness of the line
* `marker`: displays the specified marker at each data point (ex: 'o', 's', '^', '*')
* label: specifies a lebl for the plotted line, which is displayed when using `legend()`

##### Making a simple line chart

````python
# data to plot
years = range(2000, 2012) 
apple_yield = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375, 0.9372, 0.939]
orange_yield = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898, 0.9, 0.896, ]

# plot a line with years on x-axis, apple_yield on y-axis
plt.plot(years, apple_yield)
# plot a line with years on x-axis, orange_yield on y-axis
plt.plot(years, orange_yield);

# add axes labels to show what each axis represents
plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')

# add a title to the line chart
plt.title('Crop Yields in Kanto');
````

##### Adding a legend

There are multiple ways to add a legend to a line chart with multiple lines. 

**1. Pass labels during line creation using `label`**

Pass the name of each line using the `label` keyword argument directly during creation. Then, call `plt.legend()` to display them. 

````python
# plot lines and give them labels directly
plt.plot(years, apple_yield, label='Apples')
plt.plot(years, orange_yield, label='Oranges')

# call legend to display labels automatically
plt.legend()

plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto');
````

**2. Pass labels as a list using `plt.legend()`**

````python
plt.plot(years, apple_yield)
plt.plot(years, orange_yield)

plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')

# pass labels as a list in the exact order of creation
plt.legend(['Apples', 'Oranges']);
````

**Other customizations:**

You can pass optional arguments to `plt.legend()` to style or reposition the legend.

````python
plt.plot(years, apple_yield, label='Apples')
plt.plot(years, orange_yield, label='Oranges')
plt.legend()
plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto');

# change position of legend
plt.legend(loc='upper left')
# plt.show()

# arrange legend items horizontally
plt.legend(ncol=2)
# plt.show()

# remove the legend border
plt.legend(frameon=False);
# plt.show()
````

##### Adding line markers

**1. Adding markers with `marker` parameter**

We can show markers for the data points on a line using the `marker` argument of `plt.plot()`. Matplotlib supports many different types of markers (ex: circle, cross, square, diamond, etc.) - find the full list of marker types [here](https://matplotlib.org/3.1.1/api/markers_api.html). 


````python
# plot apples yield line with circle markers
plt.plot(years, apple_yield, marker='o')
# plot oranges yield line with cross markers
plt.plot(years, orange_yield, marker='x')

plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')
plt.legend(['Apples', 'Oranges']);
````

The `plt.plot()` function supports many argumets for styling lines and markers:
* `color` or `c`: set color of line
* `linestyle` or `ls`: choose between solid or dashed line
* `linewdith` or `lw`: set the width of a line
* `markersize` or `ms`: set the size of markers
* `markeredgecolor` or `mec`: set the edge color for markers
* `markeredgewidth` or `mew`: set the edge with for markers
* `markerfacecolor` or `mfc`: set the fill color for markers
* `alpha`: set the opacity of the plot

**2. Styling markers with `fmt` parameter**

The optional `fmt` parameter of `plt.plot()` is a convenient shorthand notation for specifying the marker, line style, and line color. It can be provided as the third argument to `plt.plot()`.

Syntax: `fmt = 'marker|line|color`
* `marker`: specifies the style of the points (ex: 'o' for circles, 's' for squares)
* `line`: specifies the style of the line (ex: '-' for solid, ':' for dotted)
* `color`: specifies the color of the plot elements (ex: 'r' for red, 'b' for blue)

````python
# use fmt parameter to specify 's-b': square, solid, blue markers
plt.plot(years, apple_yield, 's-b')
# use fmt parameter to specify 'o--r': circle, dashed, red markers
plt.plot(years, orange_yield, 'o--r')

plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')
plt.legend(['Apples', 'Oranges']);
````

**3. Plotting markers only without a line**

You can also display data points without any connecting lines (like a basic scatterplot) by not specifying the line style in `fmt`.

````python
# use fmt parameter to specify 'or': circle, red markers
plt.plot(years, orange_yield, 'or')
plt.title('Yield of Oranges in Kanto')

plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
````

### Scatter Plot

A scatter plot allows us to see the relationship between 2 variables by plotting their values as points on a 2-dimensional grid. Additionally, more variables can be used to change the size, color, or style of the data points. 

Let's use the Iris flower dataset included with the Seaborn library, which provides sample measurements of sepals and petals for 3 species of flowers.

````python
# load data into a Pandas dataframe
flowers_df = sns.load_dataset("iris")
flowers_df
````

````python
# the Pandas .unique() method extracts distinct values from a Series (a single column) 
# and returns them as a NumPy array
flowers_df.species.unique()
````

Let's try to visualize the relationship between sepal length and sepal width. If we tried to do so by creating a line chart using `plt.plot()`, we would find that the output is not very informative - the two variables are not related in a specific sequential order.

````python
plt.plot(flowers_df.sepal_length, flowers_df.sepal_width);
````

A line chart is mainly used to show sequential changes or trends over equal intervals of time. In contrast, a scatter plot is useful for showing the correlation or relationship between two numeric variables. 

The `sns.scatterplot()` function from `seaborn` (imported as `sns`) is used to create a scatter plot.

##### **`sns.scatterplot()`**

Syntax: 
```
sns.scatterplot(
    x=None, y=None,
    hue=None, style=None, size=None,
    data=None, palette=None,
    legend='brief', alpha='auto'
)
```

Parameters:
* `x`, `y`: Column names, or, variables for the positions on the x and y axes
* `data`: The input data structure (Pandas DataFrame)
* `hue`: Variable that groups data by color
* `size`: Variable that groups data by marker size
* `style`: Variable that groups data by marker shape (dots, crosses, squares)
* `palette`: Sets the color palette for the `hue` grouping
* `alpha`: Transparency of points
* `legend`: Controls legend visibility and behavior

##### Making a simple scatter plot

A basic scatter plot requires data for the x and y axes. Here, the column names of the columns we want to plot are passed into the x and y axes. The `data` argument takes the DataFrame that provides the column names and values.

````python
sns.scatterplot(
    # pass 'sepal_length' column values into x-axis
    x='sepal_length',
    # pass 'sepal_width' column values into y-axis
    y='sepal_width',
    # pass 'flowers_df' DataFrame for data
    data=flowers_df
);
````

##### Setting title and axes labels

The simple scatter plot above is missing a title and properly formatted x and y labels (the current ones are the column names). 

This can be accomplished using the `set_[...]()` methods, which allow you to change individual properties of a specific plot object.

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df
)

# set_() methods to specify title, axes labels
plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

##### Adding hues

The `hue` parameter is groups data by a third variable, assigning a different color to each category (each distinct value). 

To do so, pass a column name from the DataFrame to `hue`. This splits your data points by color, which is assigned depending on the value of that data point for the specified column (passed into `hue`). 

This also automatically creates a legend.

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    # group data points (assign different colors) based on species
    hue='species'
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

In conjunction, we can specify the `palette` parameter to choose a preferred color scheme. We can also set a `hue_order`, which allows you to control the sequence of categorical levels mapped by `hue`. For example, by passing a list of values to `hue_order`, you can specify the order of items appearing in the plot legend. 

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    hue='species',
    # specify palette from Seaborn palettes list
    palette='viridis',
    # passing a list of species values with custom order
    hue_order = ['setosa', 'versicolor', 'virginica']
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

*Note:* If the `hue` column contains numbers, Seaborn will switch from a discrete palette to a continuous sequential color map.

##### Setting marker style

The `style` parameter maps a categorical variable to different marker shapes. This helps group and differentiate data points visually by relying on shape rather than color. 

*Note:* This parameter will treat any numeric entries as categorical.

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    # assign different styles based on species
    style='species'
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

You can further control the markers with these related parameters:
* `markers`: A list of specific marker shapes, or, a dictionary that maps categories to shapes
* `style_order`: A list that explicitly orders how the categories are processed and mapped to shapes.
* `marker`: Applies a single uniform shape to all points (ex: `marker='v'`)

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    # assign different styles based on species
    style='species',
    # using a dict to map each species to a marker shape
    markers={'setosa': 'o', 'versicolor': 's', 'virginica': 'D'}
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

##### Setting point size

The `size` parameter maps the size of markers to a variable in the data. By passing a column name to `size`, we can make data point sizes vary based on that column's values. 

*Note:* For numeric variables, marker sizes are assigned on a continuous scale proportional to the values. For categorical variables, a distinct size is assigned to each unique category.

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    # assign different sizes based on species
    size='species',
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

You can further control the marker sizes with these related parameters:
* `sizes`: A tuple or list that defines the minimum and maximum size of the markers.
* `size_order`: A list that explicitly orders how the categories are processed and mapped to sizes.
* `s`: Applies a unform sizing to all data point marker. 

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    # assign different sizes based on species
    size='species',
    # specify min and max marker sizes
    sizes=(80, 200)
)

plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

##### Changing opacity

The `alpha` parameter controls the opacity of the plotted points. This is useful for resolving overplotting, allowing you to easily see where data points heavily overlap and cluster.

The `alpha` value accepts a float between 0 (completely invisible) and 1 (fully opaque).

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    size='species',
    sizes=(80, 200),
    # setting the opacity of data points to 0.7
    alpha=0.7
)
plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

Here, we can further improve the clarity of the plot by changing the `edgecolor` parameter (which is a Matplotlib parameter, but can be used directly in Seaborn functions). This controls the color of the borders around the plotted data points. 

The default of `edgecolor` is `white` - a white outline is applied to scatter plots when they overlap. You can pass in any standard color name, hex code, or RGB tuple to change it.  

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    size='species',
    sizes=(80, 200),
    alpha=0.7,
    # change outline color of points to black
    edgecolor='black'
)
plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

##### Customizing the legend

The `legend` parameter determines how the legend is drawn. By default, Seaborn automatically generates a  legend if any of the `hue`, `size`, or `style` semantic parameters are used. 

Parameter values:
* `brief`: Shows a sample of evenly spaced values for numeric `hue` and `size` variables.
* `full`: Every category gets an entry in the legend.
* `auto` (default): Automatically decides between `'brief'` and `'full'` depending on the number of discrete categories in the data.
* `False`: No legend is drawn.

````python
plot = sns.scatterplot(
    x='sepal_length',
    y='sepal_width',
    data=flowers_df,
    size='species',
    sizes=(80, 200),
    alpha=0.7,
    edgecolor='black',
    # draw plot without legend
    legend=False
)
plot.set_title('Sepal Length vs. Sepal Width of Iris Species')
plot.set_xlabel('Sepal Length (mm)')
plot.set_ylabel('Sepal Width (mm)');
````

### Histogram

A histogram represents the distribution of data by forming 'bins' along the range of data and plotting the number of observations that fall within discrete bins. Often, a histogram is used to show the distribution of a single variable - a single column in a dataset.

The `sns.histplot()` function is used to create a histogram.


##### **`sns.histplot()`**

Syntax: `sns.histplot(data, x, y, hue, stat, bins, bindwdth, discrete, kde, log_scale)`

Parameters:
* `data`: input data in the form of a DataFrame or Numpy array
* `x`, `y`: key (column names) of the data to be positioned on the x and y axes respectively
* `hue`: semantic data key, mapped to determine the color of plot elementsi

##### Making a simple histogram

Create a basic histogram by passing the dataset to the `data` parameter. Specify the variable (column name in Pandas DataFrame) being plotted using `x` (for a vertical histogram) or `y` (for a horizontal histogram). 

````python
# plot a histogram showing the distribution of sepal widths
plot = sns.histplot(data=flowers_df, 
                    # histogram is drawn vertically
                    x='sepal_width'
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

````python
# plot a histogram showing the distribution of sepal widths
plot = sns.histplot(data=flowers_df, 
                    # histogram is drawn horizontally
                    y='sepal_width'
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

##### Adjusting size and number of bins

The `bins` parameter allows us to control the size and number of bins - the way the data is grouped and segmented in the histogram. 

**1. Setting the number of bins**

Pass a single integer to define the number of equally spaced intervals - bins - you want.

````python
plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    # split data into 20 bins
                    bins=20
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

*Note:* Alternatively, you can pass a string reference rule recognized by NumPy (ex: `auto`, `fd`) to calculate the optimal bin count mathematically ([look into the NumPy docs for more info](https://numpy.org/devdocs/reference/generated/numpy.histogram_bin_edges.html)).

**2. Setting exact bin edges**

If you want to split data into unequal intervals, you can manually pass a list `bins`. 

````python
plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    # bins of unequal sizes
                    bins=[2, 2.7, 3, 3.3, 4, 4.6]
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

If you want to have explicitly defined boundaries for each bin, but equal intervals, you can create and pass a NumPy array to `bins`.

````python
import numpy as np

plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    # specifying bin boundaries (equally spaced) using NumPy array
                    bins=np.arange(2, 5, 0.25)
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

**3. Setting a uniform bin width**

Instead of choosing the number of bins, you can define the width of each bin using the `binwidth` parameter. This automatically computes the number of bins needed to cover the data range. 

````python
plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    # every bin will represent an interval of 0.25 units
                    binwidth=0.25
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

**4. Limiting the bin range**

You can restrict the overall boundaries between which bins are calculated using the `binrange` parameter. This is useful for filtering out outliers without altering your source DataFrame. 

````python
plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    # bins are only drawn for data between 2.0 and 4.0
                    binrange=(2, 4)
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');
````

##### Adding statistical information

You can display the mean and median of the data on the histogram distribution. This can be done using the Matplotlib `plt.axvline()` function, which adds a vertical line across the height of a plot. 

##### **`plt.axvline()`**

Syntax: `plt.axvline(x=0, ymin=0, ymax=1, **kwargs)`

Parameters:
* `x` (float): Position on x-axis where the line will be drawn, given in data coordinates
* `ymin` (float): Starting point of the line along the y-axis, given in axes coordinates (0 = bottom, 1 = top)
* `ymax` (float): Ending point of the line along the y-axis, given in axes coordinates (0 = bottom, 1 = top)
* `**kwargs`: Line styling properties passed to parameters like `color`, `linestyle`, `linewidth`, `label`

````python
import numpy as np

# compute mean and median of sepal_width column
mean = flowers_df['sepal_width'].mean()
median = flowers_df['sepal_width'].median()

plot = sns.histplot(data=flowers_df, 
                    x='sepal_width',
                    bins=np.arange(2, 5, 0.25)
)
plot.set_title("Distribution of Sepal Width")
plot.set_xlabel('Sepal Width (mm)');

# draw an orange dashed line at the mean
plt.axvline(x=mean, color='orange', label='Mean', linestyle='--', linewidth=3)
# draw a black dashed line at the median
plt.axvline(x=median, color='black', label='Median', linestyle='--', linewidth=3)
plt.legend();
````

##### Plotting multiple variables

We can plot multiple variables by drawing multiple histograms in a single chart. If we simply provide the entire dataset without assigning x or y to column names, a histogram is created for each numerical column in the dataset, with a different color for each variable (column).

To configure how multiple subsets or distributions are handled in `sns.histoplot()`, you can use the `multiple` parameter, combined with a categorical variable passed to the `hue` argument.

The `multiple` parameter accepts four keyword arguments:
* `'layer'` (default): Overlays the distributions transparently on top of each other. This is useful for checking the overlap of small data groups or for comparing the internal distribution of each subset.
* `'stack'`: Stacks the categorical groups vertically on top of each other. This is useful for visualizing the overall distribution total while seeing group breakdowns.
* `'dodge'`: Places the bars side by side next to each other within the same bin. This is useful for directly comparing counts of specific bins across groups.
* `'fill'`: Normalize all bars to 100% height, showing a proportional breakdown. This is useful for spotting changes in group ratios across the data range.

**1. Overlapping histograms (`'multiple=layer'`)**

By default, the histograms will overlap with each other. This kind of graph is useful if you want to compare the internal distribution of each individual category. We can map colors to different data groups (columns) by passing a categorical column name to the `hue` parameter. As seen, the opacity of the bars are automatically reduced.

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    # assign different colors by species
                    hue='species')

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````

We can also manually adjust the opacity of each histogram using the `alpha` parameter so that overlapping bars don't hide each other.

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    hue='species',
                    # turn down opacity
                    alpha=0.4
                    )

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````

**2. Stacking histograms (`multiple='stack'`)**

Rather than have the histograms overlapping each other, we can stack them on top of one another by assigning the keyword `'stack'` to the `multiple` parameter. This can be useful when you want to show the *total* distribution of a dataset while simultaneously revealing how individual categories contribute to that total. (Notice how the count changes on the y-axis between the previous example and this one).

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    hue='species',
                    # stack histograms on top of each other
                    multiple='stack'
                    )

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````

**3. Placing bars side by side within a bin**

By passing the keyword `'dodge'` to the `multiple` parameter, you can place the bars side by side next to each other within the same bin. This is useful for directly comparing the counts of specific bins across groups. 

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    hue='species',
                    # place bars side by side in each bin
                    multiple='dodge',
                    )

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````

**4. Changing counts to proportion**

By passing the keyword `'fill'` to the `multiple` parameter, you can normalize all bars to 100% height so that the histogram shows a proportional breakdown of each category, rather than the absolute counts.

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    hue='species',
                    # show porportional breakdown
                    multiple='fill',
                    )

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````

To make the histogram clearer, you can specify the `shrink` parameter, which accepts a value between 0 and 1 (controlling the size of the bars). This adds a small gap between bins, making the bars cleaner to read. 

````python
plot = sns.histplot(data=flowers_df,
                    x='sepal_width',
                    hue='species',
                    multiple='fill',
                    # add gap between bins by shrinking bars
                    shrink=0.9
                    )

plot.set_title('Distribution of Sepal Width')
plot.set_xlabel('Sepal Width (mm)');
````
