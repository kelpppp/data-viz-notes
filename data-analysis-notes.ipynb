#### **Methods & Properties**

##### `df.shape`




**Definition:** The `shape` property retrns a tuple containing the shape (the number of rows and columns) of the DataFrame. 

**Syntax:** `df.shape()`

**Return value:** `(num_rows, num_cols)`

**Access individual dimensions:**
* Get the number of rows only using `df.shape[0]`
* Get the number of columns only using `df.shape[1]`

##### `df.info()`

**Definition:** 

The `df.info()` method print a summary of a DataFrame. 


This information contains: 
* Total entries: The number of rows and the range index boundaries.
* Column Invetory: A list of all column names.
* Non-null counts: The number of non-missing values in each column.
* Data types (`dtypes`): The data type of the values of each column.
* Memory usage: Estimate of the RAM consumed by the DataFrame.

**Syntax**: 

* Required: `df.info()`

* Full: `df.info(verbose, buf, max_cols, memory_usage, show_counts, null_counts)`


**Parameters:**
* `verbose` (`True` or `False`): Whether to print all information or not
* `max_cols` (`int`): Number of columns you want information from
* `show_counts` (`True` or `False`): Whether to print the non-null counts or not

##### `df.describe()`

**Definition:**  The `df.describe()` method returns descriptive statistics about the data in the DataFrame. 

**Syntax:**
`df.describe(percentiles=None, include=None, exclude=None)`

**Parameters:**
* `percentiles`: A list of numbers between 0 and 1 specifying which percentiles to return. Default `None` returns the 25th, 50th, and 75th percentiles.
* `include`: A list of data types to include in the description (ex: `int`, `float`, `object` for strings). Default `None` means only numeric types are included (if DataFrame contains no numeric columns, all types are included). 
* `exclude`: A list of data types to exclude from summary. Default `None` means no types are excluded. 

**Using describe() on a numerical DataFrame:**
If the DataFrame contains numerical data, the description contains these information for each column:
* `count`: Number of non-null entries.
* `mean`: Mathematical average of values.
* `std`: Standard deviation (measures data spread).
* `min`: Lowest value in the column.
* `25%`: the 25th percentile* (first quartile).
* `50%` the 50th percentile* (the median).
* `75%`: The 75th percentile* (third quartile).
* `max`: Highest value in the columb.

*The value of a percentile is the specific number below which a given percentage of data points in a dataset fall. 

**Using describe() on other data:**
If the DataFrame contains strings, booleans, or categorical data, `describe()` returns different statistics:
* `count`: Number of non-null entries.
* `unique`: Number of distinct values.
* `top`: Most commonly occurring value (mode).
* `freq`: Freqeuncy (count) of the most common value (mode).

##### `df.head()`

**Definition:** The `df.head()` method returns the first `n` rows (if specified) of a DataFrame or Series. If `n` is not specified, it returns the first 5 rows. It is primarily used for quickly previewing and inspecting the structure of your data after loading it. 

**Syntax:** `df.head(n)`

**Parameters:**
* `n` (optional): The number of rows to retrieve from the top of the Dataframe of Series.

##### `df.tail()`

**Definition:** The `df.head()` nethod returns the last `n` rows (if specified) of a DataFrame of Series. If `n` is not specified, it returns the last 5 rows. It is primarily used for quickly previewing and inspecting the structure of your data after loading it. 

**Syntax**:
* Required: `df.tail()`
* Full: `df.tail(n)`

**Parameters:**
* `n` (optinal): The number of rows to retrieve from the bottom of the Dataframe of Series.

##### `df.sample()`

**Definition:** The `df.sample()` method is used to randomly select rows or columns from a DataFrame. Calling `df.sample()` without any argments returns **one** random row from the dataset. 

**Syntax:** 
* Required: `df.sample()`
* Full: `df.sample(n=None, frac=None, replace=False, weights=None, random_state=None, axis=None)`

**Parameters:**
* `n` (`int`): Number of random rows to generate.
* `frac` (`float`): The fraction (percentage) of items to return. *Note: you either use `n` or `frac`, not both.*
* `replace` (`boolean`): If `True`, allows the method to sample the same row more than once.
* `random_state`: (`int`): The seed for the random number generator. Setting this to the same value ensures you get the exact same random sample every time the code is run. 
* `axis`: Whether to sample rows (`axis=0` or `index`) or columns (`axis=1` or `columns`).
* `ignore_index` (`boolean`): If `True`, the index of the returned DataFrame will be reset instead of keeping original index labels. 

[More info](https://www.geeksforgeeks.org/python/python-pandas-dataframe-sample/)
